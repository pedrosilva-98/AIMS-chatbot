```python
import json
import re
import os
```

```python
def clean_netpa_content_final(text):
    if not text:
        return ""

    # Remover conteúdo desnecessário do NetPA    
    padrao_rodape = r"netPA | NOVA IMS\n Optimizado para 1440x900 | Chrome, Safari, Edge, FX, Opera\n(SmartPhones e Tablets não são suportados na totalidade)"
    
    text = re.sub(padrao_rodape, "", text, flags=re.DOTALL | re.IGNORECASE)
    
    lixo_menu = [
        "PT EN",
        "MINHA INFORMAÇÃO",
        "CANDIDATURAS",
        "ÁREA CONSULTAS",
        "ATIVIDADES LETIVAS",
        "ATIVIDADES NÃO LETIVAS",
        "SITUAÇÃO FINANCEIRA",
        "OUTROS SERVIÇOS",
        "MEU PERFIL",
        "SAIR",
        "Abrir",
        "SALTAR PARA CONTEÚDO"
    ]
    
    for item in lixo_menu:
        text = text.replace(item, "")

    text = re.sub(r'\n{3,}', '\n\n', text)
    
    return text.strip()
```

```python
input_file = "../netpa_scraping/netpa_dump.json"        
output_dir = "netpa_btns" 

os.makedirs(output_dir, exist_ok=True)

if os.path.exists(input_file):
    
    with open(input_file, 'r', encoding='utf-8') as f:
        data = json.load(f)

    cleaned_data = []
    
    for entry in data:
        original = entry.get('conteudo', '')
        
        limpo = clean_netpa_content_final(original)
        
        entry['conteudo_limpo'] = limpo
        cleaned_data.append(entry)       

    output_path = os.path.join(output_dir, "netpa_dump_limpo.json")    

    with open(output_path, 'w', encoding='utf-8') as f:
        json.dump(cleaned_data, f, ensure_ascii=False, indent=2)
```
